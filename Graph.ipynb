```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Drawing Axes (x and y)
plt.plot([i for i in range(-100, 100)],[0 for i in range(-100, 100)], color='black')
plt.plot([0 for i in range(-100, 100)],[i for i in range(-100, 100)], color='black')

# Setting axes limit
xlim_left, xlim_right = [-4, 4]
ylim_left, ylim_right = [-4, 4]
plt.xlim([xlim_left, xlim_right])
plt.ylim([ylim_left, ylim_right])

# Declaration of padding
base_padding = 0.1

# Label the axes
plt.text(xlim_right + base_padding, 0, '+x')
plt.text(xlim_left - 3*base_padding, 0, '-x')
plt.text(0, ylim_right + base_padding, '+y')
plt.text(0, ylim_left - 3*base_padding, '-y')
# plt.xlabel('x axis')
# plt.ylabel('y axis')

# Drawing the main axes equation
plt.text(base_padding, ylim_left / 2, 'x = 0')
plt.text(xlim_left / 2, base_padding, 'y = 0')
# Mark origin
plt.text(base_padding, base_padding, 'O')

# Drawing parallel of x axis
for value in range(ylim_left+1, ylim_right, 2):
    plt.plot([i for i in range(-100, 100)],[value for i in range(-100, 100)], color='blue')
    plt.text(xlim_left / 2, value + base_padding, f'y = {value}', color='blue')

# Drawing parallel of y axis
for value in range(xlim_left+1, xlim_right, 2):
    plt.plot([value for i in range(-100, 100)],[i for i in range(-100, 100)], color='green')
    plt.text(value+base_padding, ylim_left / 2, f'x = {value}', color='green')

# END GENERATE GRAPH

plt.tick_params(left = False, right = False , labelleft = False, labelbottom = False, bottom = False) 
plt.grid()
plt.show()
```
